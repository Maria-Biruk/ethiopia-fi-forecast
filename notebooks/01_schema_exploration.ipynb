```python
import pandas as pd

xls = pd.ExcelFile("../data/raw/ethiopia_fi_unified_data.xlsx")
print("Sheet names:", xls.sheet_names)

refs = pd.read_excel("../data/raw/reference_codes.xlsx")
print("Reference codes shape:", refs.shape)
refs.head()
```

```python
data_df = pd.read_excel("../data/raw/ethiopia_fi_unified_data.xlsx", sheet_name="ethiopia_fi_unified_data")
impact_df = pd.read_excel("../data/raw/ethiopia_fi_unified_data.xlsx", sheet_name="Impact_sheet")

print("Main sheet shape:", data_df.shape)
print("Main sheet columns:", list(data_df.columns))
print()
print("Impact sheet shape:", impact_df.shape)
print("Impact sheet columns:", list(impact_df.columns))
```

```python
print("Reference codes columns:", list(refs.columns))
refs.head(10)
```

```python
combined_df = pd.concat([data_df, impact_df], ignore_index=True)
print("Combined shape:", combined_df.shape)

print("\nRecords by record_type:")
print(combined_df["record_type"].value_counts(dropna=False))

print("\nRecords by pillar:")
print(combined_df["pillar"].value_counts(dropna=False))

print("\nRecords by source_type:")
print(combined_df["source_type"].value_counts(dropna=False))

print("\nRecords by confidence:")
print(combined_df["confidence"].value_counts(dropna=False))
```

```python
print(combined_df["source_type"].value_counts(dropna=False).to_string())
```

```python
print("observation_date range:")
print(pd.to_datetime(combined_df["observation_date"], errors="coerce").min(), "to",
      pd.to_datetime(combined_df["observation_date"], errors="coerce").max())

print("\nperiod_start / period_end range:")
print(pd.to_datetime(combined_df["period_start"], errors="coerce").min(), "to",
      pd.to_datetime(combined_df["period_end"], errors="coerce").max())

print("\nfiscal_year values:")
print(combined_df["fiscal_year"].dropna().unique())

print("\nHow many rows actually have an observation_date?")
print(combined_df["observation_date"].notna().sum(), "out of", len(combined_df))
```

```python
print("Unique indicator_code values and counts:")
print(combined_df["indicator_code"].value_counts(dropna=False).to_string())

print("\nFor each indicator_code, which record_types use it:")
print(combined_df.groupby("indicator_code")["record_type"].apply(lambda x: x.value_counts().to_dict()))
```

```python
coverage = pd.crosstab(combined_df["indicator_code"], combined_df["record_type"])
print(coverage.to_string())
```

```python
usage_rows = combined_df[combined_df["pillar"] == "USAGE"][
    ["record_id", "indicator", "indicator_code", "value_numeric", "observation_date", "confidence"]
]
print(usage_rows.to_string())
```

```python
events = combined_df[combined_df["record_type"] == "event"][
    ["record_id", "indicator_code", "category", "observation_date", "source_name", "confidence"]
].sort_values("observation_date")
print(events.to_string())
```

```python
impacts = combined_df[combined_df["record_type"] == "impact_link"][
    ["record_id", "parent_id", "indicator", "related_indicator", "relationship_type",
     "impact_direction", "impact_magnitude", "impact_estimate", "lag_months",
     "evidence_basis", "comparable_country", "confidence"]
]
print(impacts.to_string())
```
